## Goal: Aggregate statistics

Often, we want to compute aggregate statistics over row

The `groupby` function allows us to split the DataFrame into groups.

 - We can group data into new DataFrames
 - We can apply a method to groups of data


The most important input to this function is what we are grouping *by*, a column name. The general format of the `groupby` operation is given by:

```python
df.groupby(by='group_name')
```

````python
chicago_grouped_nb= chicago_airbnb_df.groupby('neighbourhood')
chicago_grouped_nb
````

````python
chicago_airbnb_df.groupby('neighbourhood').mean(numeric_only=True)
````

````python
chicago_airbnb_df.groupby('neighbourhood').mean(numeric_only=True)[['price']]
````

## What can I aggregate by?

**Common Aggregate Functions**

.count() – Number of observations

.sum() – Sum of values

.mean() – Mean of values

.median() – Arithmetic median of values

.min() – Minimum

.max() – Maximum

.mode() – Mode

.std() – Standard deviation

.var() – Variance

**How many AirBnBs are in each neighborhood?**

````python
chicago_airbnb_df.groupby('neighbourhood').count()
````

**What is the minimum price by neighborhood?**

````python
grouped_nb_min = chicago_airbnb_df.groupby('neighbourhood').min(numeric_only=True)[['price']]
grouped_nb_min
````

Does this change the way we extract elements?

````python
grouped_nb_min.price['Albany Park']
````

Notice we have a DataFrame with a column name as an index. If we don't want that, we can use a method
```python
.reset_index()
```

````python
grouped_nb_min.reset_index()
````

### We can group by multiple columns

Notice the row indexes are constructed from both grouping columns

````python
chicago_airbnb_df.groupby(['neighbourhood', 'room_type']).count()[['id']]
````

We would use tables like this for cross-classification - looking at patterns between variables

Let's use a **pivot_table**!

## General Pivot_table
A *pivot_table* allows for cross-classification of groups in a DataFrame.

The general format for a *pivot_table* is –

```python
df.pivot_table(values = 'data', index='group_1', columns='group_2', aggfunc='function')
```

– where:

* each unique value in `index` gets its own row

* each unique value in `columns` gets its own column

* and `data` specifies the value in the DataFrame to which we want to apply `aggfunc`

The default option for `aggfunc` is `mean`.

````python
chicago_airbnb_df.pivot_table(values = 'price', index = 'neighbourhood', columns = 'room_type', aggfunc = 'mean')
````

We can use the argument `fill_value` to replace those NaN values with 0 since no airbnb's in those groups.

````python
chicago_airbnb_df.pivot_table(values = 'price', index = 'neighbourhood', columns = 'room_type', aggfunc = 'mean', fill_value = 0)
````
