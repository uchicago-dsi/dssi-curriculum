# Functions

Often we find while programming that there is code we want to reuse.

For instance, maybe we are combining data temperature data sets some of which are in Fahrenheit and some are in Celsius. If we want to do any analysis we should do this in the same units. We might find we are rewriting the same code over and over again.

````python
# Temps in Fahrenheit
tempA = 33.1
tempB = 45.67
tempC = 212.0

# convert these to Celsius
print((tempA-32)*5/9)
print((tempB-32)*5/9)
print((tempC-32)*5/9)
````

When we find ourselves doing this it can be useful to write our own function.

````python
def FtoC(temp):
    """Converts a temperature from Fahrenheit to Celsius"""
    return (temp-32)*9/5
````

````python
print(FtoC(tempA))
print(FtoC(tempB))
print(FtoC(tempC))
````

One big advantage is if I made a mistake like I did above. I only have to correct it in one place.

````python
def FtoC(temp):
    """Converts a temperature from Fahrenheit to Celsius. Correctly this time."""
    return (temp-32)*5/9
````

````python
print(FtoC(tempA))
print(FtoC(tempB))
print(FtoC(tempC))
````

The string that appears after the function definition is the doc string. This should describe what the function does, it will be returned if you call ```help``` on the function. This is useful to you and anyone else you work with on a project.

````python
help(FtoC)
````

## Warning

A common mistake when writing a doc string, is to accidentally write your doc string as a comment. Obviously this is better than no doc string, but it's not as useful, because ``help`` won't return the documentation.

````python
def FtoC(temp):
    #Converts a temperature from Fahrenheit to Celsius. Correctly this time.
    return (temp-32)*5/9
help(FtoC)
````

## More Complicated Functions - Multi-line Functions

So far all functions we have defined consisted of just a return statement. However, most of the time we will find ourselves wanting to define a function which repeats more than a single line of code.

````python
def percents(values):
    """This function takes a array of numbers and returns the proportion of the total
    that each value takes up."""
    total = sum(values)
    proportions = values/total
    return np.round(proportions*100,decimals=2)
````

### Optional Arguments

Sometimes we want to have more flexibility when defining a function but also now loose the convenience of an earlier definition.


````python
def new_percents(values, places):
    """This function takes a array of numbers and returns the percent of the total that each value takes up.
    values = array of values
    places = number of decimal places to round percent to."""
    proportion = values/sum(values)
    return np.round(proportion * 100, places)

new_percents(np.array([1,2,3,43]),2)
````

Note: the function as written *requires* two arguments

````python
new_percents(np.array([1,2,3,43]))
````

We can make arguments optional by providing **default** values

````python
def new_percents(values, places = 2):
    """This function takes a array of numbers and returns the percent of the total that each value takes up.
    values = array of values
    places = number of decimal places to round percent to (default = 2)"""
    proportion = values/sum(values)
    return np.round(proportion * 100, places)
````

now the function can take one or two arguments

````python
my_array = np.array([1,2,3,43])
new_percents(my_array,4)
````

````python
new_percents(my_array)
````

You can reorder arguments by being explicit in the call

This is useful if you have multiple optional arguments

````python
new_percents(places=3, values=my_array)
````

## Apply - Using functions with DataFrames in Pandas

Let's define a function that replaces any numbers greater than 100 with 100

````python
def cut_off(age):
    return min(age, 100)
````

````python
print('adjusted age of 70 is:', cut_off(70))
print('adjusted age of 170 is:', cut_off(170))
````

Create a DataFrame

````python
ages = pd.DataFrame(
    {'Person': np.array(['A', 'B', 'C', 'D']),
    'Age': np.array([63, 101, 99, 102])}
)
ages
````

### Using "apply()" on a single column

Provide the function name and the column on which to apply the function

`apply()` returns a Series

````python
ages.Age.apply(cut_off)
````

Note, it is often good practice to retain the original data when you are modifying it for further analysis (so you don't lose data that may be useful in the future and you have a record of what you did.)

If you want to augment the table you need to do so explicitly

````python
ages['AgeCutoff'] = ages.Age.apply(cut_off)
ages
````

When applying functions to dataframes it is sometimes useful to use "anonymous" (unnamed) functions.   

These are called "lambda functions".

Note, lambda functions are also handy for impressing friends and family!


Here's the above example done with a lambda function. Note this has the advantage of not having to save the cutoff function as above.

````python
ages.Age.apply(lambda x: max(x,100))
````

In the above example, the `x` variable is bound to each value in the `Age` column because we called it on `ages.Age`

Note there's really very little different between lambda functions and the ordinary functions we've been defining. In fact, one could theoretically define a function using lambda functions.

````python
double = (lambda x: x*2)
double(4)
````

Though you should avoid doing this in part because you can't define doc strings for lambda functions, (lie) and you can't define multi line lambda functions (mostly true).

One limitation lambda functions do not have is single inputs.

````python
mult = (lambda x,y: x*y)
mult(3,7)
````

With lambda functions we can also have the variable bind to entire rows.  This is useful when your functions need to access multiple columns.

Note that we have to use `axis="columns"` (or `axis = 1`) so that the lambda function is applied to each row.

````python
ages['NewDiff'] = ages.apply(lambda x: x.Age - x.AgeCutoff,axis='columns')
ages
````

Description of "axis" from the python documentation
(seems a bit counter-intuitive to me!)

**axis**: *{0 or 'index', 1 or 'columns'}, default 0*
<br>
Axis along which the function is applied:
- 0 or 'index': apply function to each column.
- 1 or 'columns': apply function to each row.
