## Merge
Frequently, we will have more than one data set that we want to analyze, or we have some data we're interested in that is missing from a data set we do have. For instance, suppose we want to combine the airbnb information with information about grocery stores to get a summary of each neighborhood in Chicago.

To do this we can use the functions merge. When doing this operations, we need to give two columns which contain the same information.

````python
airbnb_count_df = chicago_airbnb_df.groupby("neighbourhood")[["name"]].count().reset_index()
airbnb_count_df = airbnb_count_df.rename(columns={"name":"num airbnbs"})
airbnb_count_df
````

````python
pd.merge(airbnb_count_df,grocery_count_df,left_on="neighbourhood",right_on="COMMUNITY AREA NAME")
````

You might notice a few things about the resulting dataframe. First, both columns "neighbourhood" and "COMMUNITY AREA NAME" are present in the data frame. Secondly, and more subtly, we actually lost data when doing this merge.

We started with 77 columns in the airbnb data, but end up with a dataset with only 75 columns.

This is in part due to the default behavior of ``pd.merge``. We see different behavior by setting the parameter ``how = "outer"``.

````python
comb_df = pd.merge(airbnb_count_df,grocery_count_df,left_on="neighbourhood",right_on="COMMUNITY AREA NAME",how="outer")
comb_df
````

## Options for Merge

In fact there are many different options for merge. To get a sense for what's happening let's look at some simple data.

````python
fruit1 = pd.DataFrame({"Name":["Apple","Papaya","Lemon"],
                       "Avg. Weight (g)":[186,275,58],
                       "Vitamin C (g/100 ml)":[27.5,61,53]})
fruit1
````

````python
fruit2 = pd.DataFrame({"Name":["Lemon","Apple","Apple","Orange"],
                       "Cost ($/lb)":[1.98,1.23,1.67,1.43],
                       "Color":["Yellow","Red","Green","Orange"]})
fruit2
````

Note if two DataFrames have a shared column with the same name, you can just specify the one column name.
The default option for merge is ``how = "inner"``. This combines the data only including rows contained in both dataframes.

````python
pd.merge(fruit1,fruit2,left_on="Name",right_on="Name")
````

````python
pd.merge(fruit1,fruit2,on="Name",how="inner")
````

````python
# Note swapping the order of the data frames only changes the order the information is displayed.
pd.merge(fruit2,fruit1,on="Name",how="inner")
````

``how = "outer"`` displays all information contained in both data frames. Any data that is missing will be set to NaN.

````python
pd.merge(fruit1,fruit2,on="Name",how="outer")
````
