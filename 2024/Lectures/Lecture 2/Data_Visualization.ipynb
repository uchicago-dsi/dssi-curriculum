## Types of Data

### Numerical
- Each value is from a fixed scale
- Typically some type of measurement or count
- Numerical ordering is meaningful
- Can have **discrete** or **continuous** data

#### Discrete:
- finite, countable - integers
- ex. age, number of kids

#### Continuous:
- infinite, take on any real number in an interval
- ex. income, height, test scores

## Types of Data

### Categorical
- Each value is from a fixed inventory
- May or may not have an ordering ([low, med, high] vs. [red, blue, green]
- Categories are mutually exclusive
- Can have **nominal** or **ordinal** data

#### Nominal
- a naming or labeling scale
- NO natural order is present
- ex. hair color, names of people

#### Ordinal
- ordered rank
- natural order is present
- ex. letter grade, customer satisfaction (very satisfied...very dissatisfied), education

### Just because the values are numbers, doesn’t mean the variable is numerical!

* Census data uses numerical value for `SEX` code (0, 1, and 2)

* It doesn’t make sense to perform arithmetic on these “numbers”, e.g. 1 - 0 or (0+1+2)/3 are nonsense here

* The variable `SEX` is still categorical, even though numbers were used for the categories

## Pandas has a method "plot" to do simple charting of dataframe data

It supports the following types of plots:

- ‘bar’ : vertical bar plot

# Plotting Numeric Data

Some good plot choices are line graphs and scatter plots

* Line plots are useful when the x-axis is ordered (e.g., timeseries data)
* Scatter plots are for comparing two dimensional data

Line graphs are the default when using `.plot(column_1,column_2)` but can also be created using `.plot.line(column_1,column_2)`

Scatter plots can be created using `.plot.scatter(column_1,column_2)`

### What went wrong?

There are multiple instances of 'Year' in `babies`

What we really wanted was the total number born in each year. We can get this using `groupby`

**Note: `groupby` gives us a dataframe so we can still use pandas `.plot`

````python
babies_grouped = babies.groupby("Year").sum()[["Number of Births"]]
babies_grouped
````

````python
babies_grouped.plot(x="Year",y="Number of Births")
plt.ylabel("Number of Births")
plt.title("Births by Year")
plt.legend().set_visible(False);
````

Remember 'Year' becomes an index when I use `groupby`. We need to reset our index.

````python
babies_grouped = babies_grouped.reset_index()
babies_grouped.plot(x="Year",y="Number of Births")
plt.ylabel("Number of Births")
plt.title("Births by Year")
plt.legend().set_visible(False);
````

This looks much better!

Remember we can also use `.plot.line()` or we can even use the argument `kind = "line"`

````python
babies_grouped.plot.line(x="Year",y="Number of Births")
plt.ylabel("Number of Births")
plt.title("Births by Year")
plt.legend().set_visible(False);
````

````python
babies_grouped.plot(x="Year",y="Number of Births", kind="line")
plt.ylabel("Number of Births")
plt.title("Births by Year")
plt.legend().set_visible(False);
````

There are many ways of doing the same thing. Pick the one that makes sense to you.

## Histogram

Chart to display the *distribution* of numerical values using bins

Distribution - the frequency of values

- Gather data into bins (numerical range)

- A bar in a histogram represents the frequency of that range

### Binning Numerical Values

Binning is counting the number of numerical values that lie within ranges called bins

* Bins are defined by their lower bounds (inclusive)
* The upper bound is the lower bound of the next bin

### We can make a histogram using one numeric variable

````python
babies.hist("Average Birth Weight (g)",color="limegreen")
plt.xlabel("Average Birth Weight (g)")
plt.ylabel("Frequency")
plt.title("Distribution of Birth Weight");
````

````python
# This would also work

babies["Average Birth Weight (g)"].hist(color="limegreen")
plt.xlabel("Average Birth Weight (g)")
plt.ylabel("Frequency")
plt.title("Distribution of Birth Weight");
````

We can use the `bins` argument to change the bin size or number of bins

````python
my_bins = np.array([2600,2800,3000,3200,3400,3600])
my_bins
````

````python
babies.hist("Average Birth Weight (g)",bins=my_bins, color="limegreen")
plt.xlabel("Average Birth Weight (g)")
plt.ylabel("Frequency")
plt.title("Distribution of Birth Weight");
````

````python
babies.hist("Average Birth Weight (g)",bins=30, color="limegreen")
plt.xlabel("Average Birth Weight (g)")
plt.ylabel("Frequency")
plt.title("Distribution of Birth Weight");
````

Smaller bins can show some interesting detials

Finer-grained view of the distribution, but may not be as smooth of a distribution/histogram

### Horizontal or Vertical Bars?

Useful for long category names


  
![](images/bar_vert.png)


````python
babies.groupby(['State']).sum()[["Number of Births"]].reset_index().plot.bar(x="State",y="Number of Births")
plt.ylabel("Number of Births")
plt.title("Births by State")
plt.legend().set_visible(False)
plt.ticklabel_format(axis="y",style='plain'); #Need to specify the axis since x is categorical
````

All of the state names are squished together...

````python
plt.rcParams['figure.figsize'] = [10, 4] # Note this will change the size of all future figures!
babies.groupby(['State']).sum()[["Number of Births"]].reset_index().plot.bar(x="State",y="Number of Births")
plt.ylabel("Number of Births")
plt.title("Births by State")
plt.legend().set_visible(False)
plt.ticklabel_format(axis="y",style='plain');
````

This may be easier to read if it were oriented horizontally...

````python
plt.rcParams['figure.figsize'] = [6, 10]
babies.groupby(['State']).sum()[["Number of Births"]].reset_index().plot.barh(x="State",y="Number of Births")
plt.ylabel("Number of Births")
plt.title("Births by State")
plt.legend().set_visible(False)
plt.ticklabel_format(axis="x",style='plain'); #Need to specify the axis since y is categorical
````

There are a lot of states, and it's hard to compare so many. Let's order them to make it easier

````python
# I don't have to resest the figure size
babies.groupby(['State']).sum()[["Number of Births"]].reset_index().sort_values("Number of Births").plot.barh(x="State",y="Number of Births")
plt.ylabel("Number of Births")
plt.title("Births by State")
plt.legend().set_visible(False)
plt.ticklabel_format(axis="x",style='plain');
````

````python
babies.groupby(['State']).sum()[["Number of Births"]].reset_index().sort_values("Number of Births", ascending = False).plot.barh(x="State",y="Number of Births")
plt.ylabel("Number of Births")
plt.title("Births by State")
plt.legend().set_visible(False)
plt.ticklabel_format(axis="x",style='plain');
````
