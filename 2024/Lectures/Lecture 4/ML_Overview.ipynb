### What do we mean by 'learn'? What is happening?


Focus on Supervised Learning

**Big picture idea**
    
We want to build a model (what is a model??)

In classification: a function that assigns labels (classes) to a given input.

Suppose X is a collection of all the features and y is the collection of labels


$$X = \begin{bmatrix}
x_{11}& x_{12} & x_{13}\\
x_{21} & x_{22} & x_{23}
\end{bmatrix}$$

$$y = \begin{bmatrix}
y_{1}\\
y_{2} 
\end{bmatrix}$$

<img src="https://raw.githubusercontent.com/SusannaLange/Data_118_images/main/DSSI_images/machine_learning_classification/model_example.png" alt="model_example" width="500">



In supervised learning, our model takes labeled data!

So we can ask: How wrong was our model? We can measure this by an **objective function** or **loss function**, which evaulates how well our model is performing.


Goal: Small error!!! 
So a loss function is a function that we want to minimize. This minimizes the error in the model.

Common loss function is Mean Squared Error (MSE): (used in numerical regression models)
    
(MSE): Given n datapoints, $$\frac{1}{n}\sum (y_i - \hat{y_i})^2,$$

where $y_i$ is the actual label and $\hat{y_i}$ is the predicted label.

```python
## Ignore this code for now!

def f(x):
    return (x**2)*np.sin(x)

def df(x):
    return (2*x)*np.sin(x)+(x**2)*np.cos(x)
k=5
th0=-6    # try different starting points:  -6, -2, 0, 4
alpha=0.02  # choosing a good learning rate is important (try also 0.002, 0.2)

#start the plot
plt.figure(figsize=(8, 6))
x=np.arange(-10,10,0.01)
plt.plot(x,(x**2)*np.sin(x),linewidth=1.5)
plt.scatter(th0,f(th0),c="red",s=100)

for i in np.arange(k):
    th0=th0-alpha*df(th0)
   # print(th0)
    plt.scatter(th0,f(th0),c="red",s=100)
```

**Global minimum:** The minimum output value of the function
    
**Local minimum:** A minimum output value of the function, relative to local surrounding.

This process of minimizing the loss function is called **training**.

We don't want to use all of the data during this training phase!

Why? Because we want to leave out data *unseen* to the model, so we can get **evaluate** or see how well our model would perform in the *wild*.
