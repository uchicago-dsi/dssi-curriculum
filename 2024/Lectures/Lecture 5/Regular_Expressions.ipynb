````python
import re
````

To use regular expressions with `re`, we first need to compile the patterns into objects

Then, we can use those objects to search for matches

The method `.search(str)` will search the supplied string argument for the first substring matching the pattern object

````python
p = re.compile('[a-z]')
p
````

````python
p.search("6 of Diamonds")
````

````python
p = re.compile('[a-z]+')
````

````python
p.search("6 of Diamonds")
````

If there is no matching substring, nothing is returned

````python
p.search("12345")
````

````python
print(p.search("12345"))
````

We can use a few methods to get information from these `re.Match` objects

````python
p = re.compile("[w]{3}[\.][^\.]+[\.].+")
my_search = p.search("www.datascience.uchicago.edu")
my_search # We can get each of these attributes out of my_search using the methods below
````

````python
my_search.group()
````

````python
my_search.start()
````

````python
my_search.end()
````

````python
my_search.span()
````

`.search()` is not the only way to find matching substrings

Here are a few more methods and what they are used for: 

- `.match()` - works similarly to `.search()`, but only looks at the beginning of a string
- `.findall()` - works similarly to `.search()` but returns a list of all matches (`.search()` only returns the first match)
- `.finditer()` - works similarly to `.findall()` but returns the matches as an iterator object

What is an iterator object? - Not important for now, but if you need to know later [click here](https://www.w3schools.com/python/python_iterators.asp)

````python
p = re.compile('[a-z]+')
p.match("6 of Diamonds")
````

````python
p.match("Draw a 6 of Diamonds")
````

Why is there still no match here?

````python
p = re.compile('[a-z]+',re.IGNORECASE)
my_match = p.match("Draw a 6 of Diamonds")
my_match
````

````python
my_match.group()
````

````python
my_match.span()
````

````python
my_all = p.findall("6 of Diamonds") # returns a list, so .span(), .group(), etc won't work!
my_all
````

def is_valid_email(email):
    pattern = re.compile('^[\w\._]+@[\w\._]+\.[a-z]{3}$')
    return re.search(pattern, email) != None

```python
def extract_phone_numbers(text):
    pattern = re.compile('\d{3}[-.\s]\d{3}[-.\s]\d{4}')
    return re.findall(pattern, text)
```
