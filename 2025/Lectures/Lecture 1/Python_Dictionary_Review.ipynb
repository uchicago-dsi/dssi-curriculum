##Dictionaries

Built-in data structure to store a collection of data. A dictionary is a set of `key: value` pairs where each key is unique.

We can create a dictionary with curly brackets "{}"

Entries of a dictionary are of the form `key: value`

#### Creating a Dictionary

````python
my_dict = {}
print("Empty Dictionary: ")
print(my_dict)
````

````python
my_dict = {
  1: 'Python',
  2: 'Java',
  3: 'Scala'
}
print(my_dict)
````

#### Creating a dictionary with list comprehension

````python
list_of_tuples = [('k1','v1'),('k2','v2'),('k3','v3')]
new_dict = {key: val for (key, val) in list_of_tuples}
print(new_dict)
````

#### Accessing a dictionary

We can use keys that have been used to index a dictionary to access its value.

Values can be of any datatype: `string`, `list` or even class objects.

````python
my_dict[1]
````

#### Adding an element

````python
my_dict[4] = 'Ruby'
print(my_dict)
````

#### Deleting an element

````python
del(my_dict[3])
````

#### What happens when you try to access a key that does not exist?

````python
print(my_dict[3])
````

````python
#Easy to avoid this when iterating through a dict
key = 3
if key in my_dict:
  print(my_dict[key])
````

#### List of keys

````python
# Returns a list containing dictionary’s keys
my_dict.keys()
````

#### List of values

````python
#  Returns a list of all the values of dictionary
my_dict.values()
````

#### List of items

````python
# Returns a list containing a tuple for each key value pair
my_dict.items()
````

#### Updating your dictionary

````python
# Updates dictionary with specified key-value pairs
dict2 = {
    3: 'Java',
    5: 'C',
    6: 'C++',
}
my_dict.update(dict2)
print(my_dict)
````

#### Updating a dictionary with `map()` and `lambda` function

````python
list_of_tuples = [('k1',1),('k2',2),('k3',3)]
old_dict = {key:val for (key,val) in list_of_tuples}
print('Initially, ',old_dict)
print()

new_list_of_tuples = map(lambda item: (item[0], item[1]+1), list_of_tuples)
new_dict = {key:val for (key, val) in new_list_of_tuples}
print('After change, ',new_dict)
print()
````

#### OrderedDict

Dictionaries are unordered but `OrderedDict` provides a mechanism to arrange items in order.

````python
from collections import OrderedDict
print(my_dict.items())
dict1 = OrderedDict(sorted(my_dict.items()))
print(dict1.items())
````

#### Counter

It is a collection where elements are stored as dictionary keys and their counts are stored as dictionary values. Counts are allowed to be any integer value including zero or negative counts.

````python
from collections import Counter

counter = Counter(['a', 'a', 'b'])
print(counter)
print()

#Adding elements
print("After updating the counter by adding another count of a recurring element:")
counter.update('b')
print(counter)
````

#### Exercise : Building a Vocabulary with word-count

From a corpus provided as a list of text, build a vocabulary of words with their respective word counts. Note that we only consider words with a count more than 1 in our vocabulary.

````python
corpus = ["Fears for T N pension after talks Unions representing workers at Turner Newall say they are disappointed after talks with stricken parent firm Federal Mogul.",
          "The Race is On. Second Private Team Sets Launch Date for Human Spaceflight. A second team of rocketeers competing for the ten million Ansari X Prize, a\
           contest for privately funded suborbital space flight, has officially announced the first launch date for its manned rocket.",
          "Ky. Company Wins Grant to Study Peptides. A company founded by a chemistry researcher at the University of Louisville won a grant to develop a method of \
          producing better peptides, which are short chains of amino acids, the building blocks of proteins."
          ]

print('Building vocab...')

````

<details>

<summary>Solution</summary>

```python
word_counter = Counter()
for text in corpus:
  tokens = text.split()
  word_counter.update(tokens)
print("Word Counter:")
print(word_counter)
print()

print("Vocabulary:")
vocabulary = {x: count for x, count in word_counter.items() if count>=2}
print(vocabulary)
```
